# **Understanding Figure and Axes**
---
## In Matplotlib, there are two key concepts you need to understand:

## **1.  `Figure` (The Overall Container) 🪟**
###     The Figure is essentially the entire "canvas" or "window" where your plots will be displayed.
###     You can think of it as a sheet of paper on which you're going to draw.
###     A Figure can contain one or more `Axes`.

---

## **2.  `Axes` (The Plotting Area) 📊**
###     The Axes is the actual plotting area(s) within the Figure. This is where your data is drawn as lines, bars, points, etc.
###     Each Axes has its own X and Y axes, and you can define separate titles, labels, and scales for it.
###     If you consider the Figure as a sheet of paper, the Axes is the actual "graph" or "plot" you draw on that paper.
---
---

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
# Create a Figure and a set of Axes
# plt.subplots() returns a Figure and one Axes (by default).

fig, ax = plt.subplots(figsize=(8, 5)) # figsize sets the dimensions of the Figure in inches.

# Plot data on the Axes
x = np.linspace(0, 10, 100) # 100 points between 0 and 10
y = np.sin(x) # Sine function for y

ax.plot(x, y, label='Sine Function', color='blue', linestyle='--') # Plot line on Axes
ax.set_title('Simple Plot with Matplotlib (Object-Oriented)') # Axes title
ax.set_xlabel('X-Axis') # X-axis label
ax.set_ylabel('Y-Axis') # Y-axis label
ax.legend() # Display the legend
ax.grid(True) # Show grid

# Display the Figure
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

np.random.seed(42)

x_scatter = np.random.rand(50) * 10
y_scatter = np.random.rand(50) * 10 + x_scatter
colors = np.random.rand(50)
sizes = np.random.rand(50) * 800 + 100

labels = ['A', 'B', 'C', 'D']
group1_data = np.random.randint(10, 80, 4)
group2_data = np.random.randint(10, 80, 4)
x = np.arange(len(labels))
width = 0.35

years = np.arange(2010, 2016)
data_comp1 = np.random.rand(6) * 100
data_comp2 = np.random.rand(6) * 100
data_comp3 = np.random.rand(6) * 100

matrix_data = np.random.rand(10, 10) * 100

fig, axes = plt.subplots(2, 2, figsize=(12, 9))

scatter = axes[0, 0].scatter(x_scatter, y_scatter, c=colors, s=sizes, cmap='viridis', alpha=0.7, edgecolors='w', linewidth=0.5)
axes[0, 0].set_title('Scatter Plot with Color & Size')
axes[0, 0].set_xlabel('Feature 1')
axes[0, 0].set_ylabel('Feature 2')
fig.colorbar(scatter, ax=axes[0, 0], label='Value')

axes[0, 1].bar(x - width/2, group1_data, width, label='Group 1', color='skyblue')
axes[0, 1].bar(x + width/2, group2_data, width, label='Group 2', color='lightcoral')
axes[0, 1].set_title('Grouped Bar Chart')
axes[0, 1].set_ylabel('Scores')
axes[0, 1].set_xticks(x)
axes[0, 1].set_xticklabels(labels)
axes[0, 1].legend()

axes[1, 0].stackplot(years, data_comp1, data_comp2, data_comp3, labels=['Component 1', 'Component 2', 'Component 3'], alpha=0.8)
axes[1, 0].set_title('Stacked Area Plot: Trends Over Time')
axes[1, 0].set_xlabel('Year')
axes[1, 0].set_ylabel('Total Value')
axes[1, 0].legend(loc='upper left')

heatmap = axes[1, 1].imshow(matrix_data, cmap='magma', origin='lower', aspect='auto')
axes[1, 1].set_title('Heatmap: Matrix Visualization')
axes[1, 1].set_xticks(np.arange(matrix_data.shape[1]))
axes[1, 1].set_yticks(np.arange(matrix_data.shape[0]))
axes[1, 1].set_xticklabels(np.arange(1, matrix_data.shape[1] + 1))
axes[1, 1].set_yticklabels(np.arange(1, matrix_data.shape[0] + 1))
fig.colorbar(heatmap, ax=axes[1, 1], label='Intensity')

plt.tight_layout(rect=[0, 0.03, 1, 0.95])
fig.suptitle('Matplotlib\'s Multi-Panel Power: Diverse Visualizations', fontsize=18, fontweight='bold', color='darkblue')

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
%matplotlib inline  # A good practice in Jupyter for plots to appear inline
```

# **1. Line Plot 〰️**
## Line plots are ideal for visualizing **trends over time or continuous relationships between two variables.**

```python
# Data for Line Plot
x_line = np.linspace(0, 10, 50)
y_line = np.sin(x_line)

# Create a Figure and an Axes object
fig_line, ax_line = plt.subplots(figsize=(8, 4))

# Plot the data
ax_line.plot(x_line, y_line, color='blue', linestyle='-', linewidth=2, label='Sine Wave')

# Add titles and labels
ax_line.set_title('Simple Line Plot of a Sine Wave')
ax_line.set_xlabel('X-axis')
ax_line.set_ylabel('Y-axis')
ax_line.legend()
ax_line.grid(True, linestyle=':', alpha=0.7)

plt.show()
```

# **2. Scatter Plot 🔴**
## Scatter plots are excellent for exploring **the relationship or correlation between two numerical variables and identifying clusters or outliers.**


```python
# Data for Scatter Plot
np.random.seed(0) # For reproducibility
x_scatter = np.random.rand(50) * 10
y_scatter = 2 * x_scatter + np.random.randn(50) * 2 + 5

# Create a Figure and an Axes object
fig_scatter, ax_scatter = plt.subplots(figsize=(8, 4))

# Plot the data
ax_scatter.scatter(x_scatter, y_scatter, color='green', marker='o', s=60, alpha=0.7, edgecolors='black')

# Add titles and labels
ax_scatter.set_title('Simple Scatter Plot')
ax_scatter.set_xlabel('Variable 1')
ax_scatter.set_ylabel('Variable 2')
ax_scatter.grid(True, linestyle=':', alpha=0.7)

plt.show()
```

# **3. Histogram 📊**
## Histograms are used to visualize the **distribution** of a single numerical variable, showing the **frequency** of data points within different intervals (bins).


```python
# Data for Histogram
data_hist = np.random.normal(loc=10, scale=2, size=1000) # Mean 10, std dev 2, 1000 points

# Create a Figure and an Axes object
fig_hist, ax_hist = plt.subplots(figsize=(8, 4))

# Plot the histogram
ax_hist.hist(data_hist, bins=20, color='purple', edgecolor='black', alpha=0.7)

# Add titles and labels
ax_hist.set_title('Histogram of Sample Data Distribution')
ax_hist.set_xlabel('Value')
ax_hist.set_ylabel('Frequency')
ax_hist.grid(axis='y', linestyle=':', alpha=0.7)

plt.show()
```

# **4. Bar Plot 🧮**
# Bar plots are used to **compare** values across different categories or to show the **magnitude** of each category.


```python
# Data for Bar Plot
categories = ['Product A', 'Product B', 'Product C', 'Product D']
sales = [150, 220, 180, 250]

# Create a Figure and an Axes object
fig_bar, ax_bar = plt.subplots(figsize=(8, 4))

# Plot the bar chart
ax_bar.bar(categories, sales, color=['skyblue', 'lightcoral', 'lightgreen', 'gold'], edgecolor='black', alpha=0.8)

# Add titles and labels
ax_bar.set_title('Monthly Sales by Product')
ax_bar.set_xlabel('Product Category')
ax_bar.set_ylabel('Sales Volume')
ax_bar.grid(axis='y', linestyle=':', alpha=0.7)
ax_bar.set_axisbelow(True) # Ensure grid lines are behind bars

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

x_data = np.linspace(0, 10, 25)
y1 = np.sin(x_data) + 1
y2 = np.cos(x_data)
y3 = np.sin(x_data * 1.5) - 1
y4 = np.cos(x_data * 0.8) + 2

fig, ax = plt.subplots(figsize=(9, 5))

ax.plot(x_data, y1, color='red', linestyle='-', linewidth=2, marker='o', markersize=7, label='Series 1 (Solid)')
ax.plot(x_data, y2, color='orange', linestyle='--', linewidth=1.5, marker='s', markersize=6, alpha=0.7, label='Series 2 (Dashed, Transparent)')
ax.plot(x_data, y3, color='green', linestyle='-.', linewidth=2.5, marker='^', markersize=8, markeredgecolor='black', label='Series 3 (Dash-Dot)')
ax.plot(x_data, y4, color='purple', linestyle=':', linewidth=1.8, marker='D', markersize=7, markerfacecolor='yellow', label='Series 4 (Dotted)')

ax.set_title('Matplotlib Line & Marker Properties', fontsize=14)
ax.set_xlabel('X-Value')
ax.set_ylabel('Y-Value')
ax.legend(loc='upper right')
ax.grid(True, linestyle=':', alpha=0.6)

plt.show()
```

# **Matplotlib: Precision & Publication ⚙️🔬**
## This section explores how to exert fine-grained control over your Matplotlib plots, moving **beyond** basic aesthetics to create **highly customized and publication-quality figures**. We'll cover **line properties, advanced subplot arrangements, and saving plots for various uses.**


```python
import matplotlib.pyplot as plt
import numpy as np
%matplotlib inline
```

# **1. Line Properties: Customizing Your Visuals 🌈**
## Matplotlib offers extensive control over how lines and markers appear. This allows you to **differentiate** data series clearly and **emphasize specific points.**


```python
# Data for line properties demonstration
x_prop = np.linspace(0, 10, 20)
y_prop1 = np.sin(x_prop)
y_prop2 = np.cos(x_prop)
y_prop3 = np.sin(x_prop) * np.exp(-x_prop/5) # Damped sine wave

fig_props, ax_props = plt.subplots(figsize=(10, 6))

# Line 1: Basic solid blue line with circular markers
ax_props.plot(x_prop, y_prop1,
              color='blue',           # Line color
              linestyle='-',          # Solid line
              linewidth=2,            # Line thickness
              marker='o',             # Circle marker
              markersize=8,           # Marker size
              label='Sine Wave (Solid Blue)')

# Line 2: Dashed red line with square markers and transparency
ax_props.plot(x_prop, y_prop2,
              color='red',
              linestyle='--',
              linewidth=1.5,
              marker='s',             # Square marker
              markersize=7,
              alpha=0.6,              # Transparency
              label='Cosine Wave (Dashed Red, Alpha 0.6)')

# Line 3: Dash-dot green line with diamond markers and larger size
ax_props.plot(x_prop, y_prop3,
              color='#2ca02c',        # Hex color code for green
              linestyle='-.',
              linewidth=2.5,
              marker='D',             # Diamond marker
              markersize=9,
              markeredgecolor='black', # Marker border color
              markerfacecolor='lightblue', # Marker fill color
              label='Damped Sine (Dash-Dot Green)')

ax_props.set_title('Matplotlib Line & Marker Properties Demonstration')
ax_props.set_xlabel('X-axis')
ax_props.set_ylabel('Y-axis')
ax_props.legend()
ax_props.grid(True, linestyle=':', alpha=0.6)

plt.show()
```

# **2.1 Subplots: Arranging Multiple Plots 🔄️**
## When you need to **display several related plots in a single figure**, `subplots` is your go-to tool. It allows for **clear comparisons and a comprehensive view of your data.**


```python
# Data for subplots
x_sub = np.linspace(0, 5, 50)
y_sub1 = np.exp(x_sub)
y_sub2 = x_sub**2
y_sub3 = np.random.rand(50) * 10
y_sub4 = np.log(x_sub + 1)

# Create a 2x2 grid of subplots
fig_sub, axes_sub = plt.subplots(2, 2, figsize=(12, 10))

# Plot 1: Exponential Growth
axes_sub[0, 0].plot(x_sub, y_sub1, color='darkorange')
axes_sub[0, 0].set_title('Exponential Growth')
axes_sub[0, 0].set_xlabel('Time')
axes_sub[0, 0].set_ylabel('Value')
axes_sub[0, 0].grid(True, linestyle=':', alpha=0.6)

# Plot 2: Quadratic Relationship
axes_sub[0, 1].plot(x_sub, y_sub2, color='darkviolet')
axes_sub[0, 1].set_title('Quadratic Relationship')
axes_sub[0, 1].set_xlabel('Input')
axes_sub[0, 1].set_ylabel('Output')
axes_sub[0, 1].grid(True, linestyle=':', alpha=0.6)

# Plot 3: Random Data Distribution (Scatter)
axes_sub[1, 0].scatter(x_sub, y_sub3, color='forestgreen', alpha=0.7)
axes_sub[1, 0].set_title('Random Data Scatter')
axes_sub[1, 0].set_xlabel('Index')
axes_sub[1, 0].set_ylabel('Random Value')
axes_sub[1, 0].grid(True, linestyle=':', alpha=0.6)

# Plot 4: Logarithmic Trend
axes_sub[1, 1].plot(x_sub, y_sub4, color='sienna')
axes_sub[1, 1].set_title('Logarithmic Trend')
axes_sub[1, 1].set_xlabel('Input')
axes_sub[1, 1].set_ylabel('Log Value')
axes_sub[1, 1].grid(True, linestyle=':', alpha=0.6)

# Adjust layout to prevent overlapping
plt.tight_layout(rect=[0, 0.03, 1, 0.95])
fig_sub.suptitle('Multiple Plots in One Figure (2x2 Grid)', fontsize=16, fontweight='bold', color='navy')

plt.show()
```

# **2.2 Advanced Subplot Layouts with `fig.add_subplot()` ↕️↔️**
## For **more complex or non-uniform subplot arrangements**, `fig.add_subplot()` offers **granular control** over subplot placement. It uses a **grid specification** (rows, columns, plot_number).


```python
# Example of a more complex subplot layout (e.g., one large plot and two smaller ones)
fig_custom = plt.figure(figsize=(12, 6))

# Large plot on the left (occupies 1st and 2nd rows, 1st column of a 2x2 conceptual grid)
ax1 = fig_custom.add_subplot(1, 2, 1) # 1 row, 2 columns, 1st plot
ax1.plot(x_sub, y_sub1, color='darkblue', linewidth=2)
ax1.set_title('Main Trend (Large Plot)')
ax1.set_xlabel('X')
ax1.set_ylabel('Y')
ax1.grid(True, linestyle=':', alpha=0.6)

# Small plots on the right (occupy 2nd column of a 2x2 conceptual grid, split vertically)
ax2 = fig_custom.add_subplot(2, 2, 2) # 2 rows, 2 columns, 2nd plot (top right)
ax2.scatter(x_sub, y_sub3, color='darkgreen', alpha=0.8)
ax2.set_title('Sub-Analysis 1')
ax2.set_xticks([]) # Remove ticks for cleaner look
ax2.set_yticks([])

ax3 = fig_custom.add_subplot(2, 2, 4) # 2 rows, 2 columns, 4th plot (bottom right)
ax3.hist(data_hist, bins=15, color='darkred', alpha=0.8)
ax3.set_title('Sub-Analysis 2')
ax3.set_xticks([])
ax3.set_yticks([])

plt.tight_layout()
fig_custom.suptitle('Custom Subplot Layout with add_subplot()', fontsize=16, fontweight='bold', color='navy')

plt.show()
```

# 1. **Why Seaborn? A Visual Comparison 💥**
## Let's compare a simple plot generated with default Matplotlib settings versus the same plot generated with Seaborn. Notice the difference in **aesthetics and the simplicity of the code.**


```python
# Generate sample data for comparison
np.random.seed(42)
data_compare = np.random.normal(loc=0, scale=1, size=500) # Normally distributed data
```

```python
# --- Matplotlib Default Histogram ---
fig_mpl, ax_mpl = plt.subplots(figsize=(7, 4))
ax_mpl.hist(data_compare, bins=20, edgecolor='black')
ax_mpl.set_title('Matplotlib Default Histogram')
ax_mpl.set_xlabel('Value')
ax_mpl.set_ylabel('Frequency')
plt.show()
```

# **Now, let's create a similar plot using Seaborn. 💥**
## Notice how Seaborn **automatically** adds a **Kernel Density Estimate (KDE)** and provides a **more polished look** with less explicit styling.


```python
fig_sns, ax_sns = plt.subplots(figsize=(7, 4))
sns.histplot(data_compare, bins=20, kde=True, color='skyblue', edgecolor='black', ax=ax_sns)
ax_sns.set_title('Seaborn Enhanced Histogram (with KDE)')
ax_sns.set_xlabel('Value')
ax_sns.set_ylabel('Frequency')
plt.show()
```

# **Key Takeaways from the Comparison:**

```python
import plotly.express as px
import pandas as pd
import numpy as np

# --- Simulate data for Shiraz, Iran with realistic busy spots ---

# Approximate central coordinates for Shiraz
shiraz_lat_center = 29.60
shiraz_lon_center = 52.54

# Define known busy/popular locations in Shiraz with approximate coordinates
# These are chosen to represent diverse types of busy areas (historical, commercial, recreational)
busy_spots = {
    "Vakil Bazaar": {"lat": 29.617, "lon": 52.538},
    "Hafezieh": {"lat": 29.605, "lon": 52.558},
    "Eram Garden": {"lat": 29.630, "lon": 52.528},
    "Chamran Blvd": {"lat": 29.620, "lon": 52.490},
    "Khalij Fars Complex": {"lat": 29.690, "lon": 52.420},
    "Zeytoon Shopping Center": {"lat": 29.620, "lon": 52.510}
}

num_locations = 200 # More locations for a denser map
np.random.seed(42) # for reproducibility

simulated_data = []

# Generate data points, concentrating some around busy spots
for i in range(num_locations):
    location_name = f"Location {i+1}"

    # Randomly pick a busy spot to cluster around for a subset of points
    if np.random.rand() < 0.6: # 60% chance to be near a busy spot
        chosen_spot_name = np.random.choice(list(busy_spots.keys()))
        spot_coords = busy_spots[chosen_spot_name]
        lat = spot_coords["lat"] + np.random.normal(0, 0.01) # Small deviation
        lon = spot_coords["lon"] + np.random.normal(0, 0.01) # Small deviation

        # Higher car hours and more likely to be peak hour for busy spots
        car_hours = np.random.uniform(30, 80) # Higher range
        peak_hour = np.random.choice(['Morning Peak', 'Evening Peak'], p=[0.5, 0.5]) # More likely peak
        location_name = f"{chosen_spot_name} Area {np.random.randint(1,5)}" # Give a name related to busy spot
    else: # Remaining 40% are more spread out and less busy
        lat = shiraz_lat_center + (np.random.rand() - 0.5) * 0.2 # Wider spread
        lon = shiraz_lon_center + (np.random.rand() - 0.5) * 0.2 # Wider spread
        car_hours = np.random.uniform(5, 35) # Lower range
        peak_hour = np.random.choice(['Morning Peak', 'Evening Peak', 'Off-Peak'], p=[0.2, 0.2, 0.6]) # More likely off-peak

    simulated_data.append([lat, lon, car_hours, peak_hour, location_name])

shiraz_df = pd.DataFrame(simulated_data, columns=[
    'centroid_lat', 'centroid_lon', 'car_hours', 'peak_hour', 'location_name'
])

# Define a custom color sequence for peak hours
color_map = {
    'Morning Peak': '#FFD700', # Gold
    'Evening Peak': '#FF4500', # OrangeRed
    'Off-Peak': '#1E90FF'     # DodgerBlue
}

# Create an interactive scatter plot on a map using Plotly Express
fig = px.scatter_mapbox(shiraz_df,
                        lat="centroid_lat",
                        lon="centroid_lon",
                        color="peak_hour", # Color by peak hour usage
                        size="car_hours", # Size by car hours
                        zoom=11, # Initial zoom level, adjust as needed
                        mapbox_style="open-street-map", # Open Street Map style
                        hover_name="location_name", # Show location name on hover
                        hover_data=["peak_hour", "car_hours"], # Additional data on hover
                        title="Interactive Carshare Usage Map: Shiraz, Iran (Simulated Peak Areas)", # title
                        labels={"car_hours": "Car Hours (per day)", "peak_hour": "Peak Hour Usage"},
                        color_discrete_map=color_map, # Apply custom colors
                        height=600)

# set layout to center the map on Shiraz and set margins
fig.update_layout(
    margin={"r":0,"t":50,"l":0,"b":0},
    mapbox_center={"lat": shiraz_lat_center, "lon": shiraz_lon_center},
    mapbox_zoom=11)

fig.show()
```

```python
import plotly.express as px
import pandas as pd
import numpy as np

# Load the built-in 'gapminder' dataset (original data up to 2007)
# This dataset contains historical data for life expectancy, GDP per capita, and population for various countries.
gapminder_original = px.data.gapminder()

# --- Data Extrapolation to 2025 for a more realistic and extended trend ---
# We aim to extend the time series data from 2007 up to 2025 to visualize more recent trends.
# The extrapolation is done in 2-year steps for smoother animation transitions.

# Get the last available year's data (2007) from the original dataset.
data_2007 = gapminder_original[gapminder_original['year'] == 2007].copy()

# Define the years for which we will extrapolate data.
# This creates a sequence of years from 2007 up to and including 2025, with a step of 2 years.
extrapolation_years = np.arange(2007, 2026, 2).tolist()
# Ensure that 2025 is explicitly included in case the step size doesn't land exactly on it.
if 2025 not in extrapolation_years:
    extrapolation_years.append(2025)
# Remove any duplicate years and sort the list to maintain order.
extrapolation_years = sorted(list(set(extrapolation_years)))

# Initialize an empty list to store all extrapolated dataframes.
all_extrapolated_data = []

# Loop through each extrapolation year to generate new data points.
for year_extrap in extrapolation_years:
    # Skip the year 2007 as its data is already present in the original dataset.
    if year_extrap == 2007:
        continue

    # Create a copy of the 2007 data to modify for the current extrapolation year.
    current_year_data = data_2007.copy()
    # Assign the current extrapolation year to the 'year' column.
    current_year_data['year'] = year_extrap

    # Calculate the number of years passed since 2007 for growth calculation.
    years_diff = year_extrap - 2007

    # Apply a simple growth/increase logic to GDP per capita, life expectancy, and population.
    # These are random uniform factors to simulate varied growth across countries, adjusted for the time difference.
    # GDP per Capita: Assumes an average annual growth rate (e.g., 1-2.5% per 2 years).
    current_year_data['gdpPercap'] = current_year_data['gdpPercap'] * (1 + np.random.uniform(0.01, 0.025, len(data_2007)) * (years_diff / 2))
    # Life Expectancy: Assumes a slower, additive increase (e.g., 0.05-0.15 years per 2 years).
    current_year_data['lifeExp'] = current_year_data['lifeExp'] + np.random.uniform(0.05, 0.15, len(data_2007)) * (years_diff / 2)
    # Population: Assumes an average annual growth rate (e.g., 0.25-0.75% per 2 years).
    current_year_data['pop'] = current_year_data['pop'] * (1 + np.random.uniform(0.0025, 0.0075, len(data_2007)) * (years_diff / 2))

    # Add the newly generated data for the current extrapolation year to the list.
    all_extrapolated_data.append(current_year_data)

# Concatenate the original Gapminder data with all the newly extrapolated dataframes.
# `ignore_index=True` resets the index of the combined DataFrame.
gapminder_extended = pd.concat([gapminder_original] + all_extrapolated_data, ignore_index=True)

# Ensure the 'year' column is of integer type, which is crucial for `animation_frame` in Plotly Express.
gapminder_extended['year'] = gapminder_extended['year'].astype(int)

# Create an interactive, animated 3D Scatter Plot using Plotly Express.
# This plot visualizes the evolution of countries across three key development indicators.
fig = px.scatter_3d(gapminder_extended,
                    x="gdpPercap",          # X-axis represents GDP per Capita.
                    y="lifeExp",            # Y-axis represents Life Expectancy.
                    z="pop",                # Z-axis represents Population.
                    color="continent",      # Points are colored based on their respective continent.
                    size="pop",             # The size of each point is scaled by its population, providing visual emphasis.
                    animation_frame="year", # The plot animates over the 'year' column, creating a time-series visualization.
                    animation_group="country", # Ensures smooth transitions for each country across animation frames.
                    hover_name="country",   # Displays the country name when hovering over a point.
                    # Formats additional data shown on hover: GDP (no decimals), LifeExp (1 decimal), Pop (shortened format).
                    hover_data={"gdpPercap": ':.0f', "lifeExp": ':.1f', "pop": ':.0s'},
                    log_x=True,             # Uses a logarithmic scale for the X-axis (GDP per Capita) for better distribution visibility.
                    size_max=80,            # Sets the maximum size for the markers to prevent them from becoming too large.
                    # Defines the fixed range for the X-axis to ensure consistent view across animation frames,
                    # adjusted to better show lower GDP values.
                    range_x=[50, 150000],
                    # Defines the fixed range for the Y-axis (Life Expectancy).
                    range_y=[20, 95],
                    # Defines the fixed range for the Z-axis (Population), adjusted to better show smaller countries.
                    range_z=[10000, 2000000000],
                    # Sets the main title of the plot.
                    title="Global Development Trends: Life Expectancy, GDP, and Population Over Time (3D)",
                    # Customizes axis labels for clarity.
                    labels={"gdpPercap": "GDP per Capita (Log Scale)",
                            "lifeExp": "Life Expectancy (Years)",
                            "pop": "Population"},
                    height=700) # Sets the height of the plot in pixels for a better 3D view.

# Update the layout of the figure for better presentation and initial camera angle.
fig.update_layout(
    # Configures the 3D scene camera's initial orientation.
    scene_camera=dict(
        up=dict(x=0, y=0, z=1),    # Defines the 'up' direction in the 3D scene.
        center=dict(x=0, y=0, z=0), # Defines the center point around which the camera orbits.
        # Sets the initial position of the camera's eye, adjusted to provide a better view
        # of countries with lower GDP and Life Expectancy.
        eye=dict(x=2.0, y=0.5, z=0.8)
    ),
    # Sets the margins around the plot (right, top, left, bottom) to zero for maximum plot area.
    margin={"r":0,"t":50,"l":0,"b":0},
    # Ensures the legend is displayed.
    showlegend=True,
    # Refines the animation speed for smoother observation of changes over time.
    transition=dict(duration=1500, easing="linear")
)

# Displays the interactive Plotly figure.
fig.show()
```

```python
# Initialize the Dash app
app_dash_example = dash.Dash(__name__)

# Define the layout of the app
app_dash_example.layout = html.Div([
    html.H1("Restaurant Tips Analysis Dashboard", style={'textAlign': 'center', 'color': '#2c3e50', 'fontFamily': 'Arial, sans-serif'}),

    html.Div([
        html.Label("Select Day of Week:", style={'fontWeight': 'bold', 'marginRight': '10px'}),
        dcc.Dropdown(
            id='day-dropdown',
            options=[{'label': day, 'value': day} for day in tips_dash_app_df['day'].unique()],
            value='Thur', # Default value
            clearable=False,
            style={'width': '200px'}
        ),
        html.Label("Smoker Status:", style={'fontWeight': 'bold', 'marginLeft': '20px', 'marginRight': '10px'}),
        dcc.RadioItems(
            id='smoker-radio',
            options=[
                {'label': 'All', 'value': 'All'},
                {'label': 'Yes', 'value': 'Yes'},
                {'label': 'No', 'value': 'No'}
            ],
            value='All', # Default value
            inline=True,
            style={'marginTop': '5px'}
        )
    ], style={'display': 'flex', 'justifyContent': 'center', 'alignItems': 'center', 'marginBottom': '30px', 'flexWrap': 'wrap'}),

    html.Div([
        dcc.Graph(id='scatter-plot-dash', style={'width': '50%', 'display': 'inline-block', 'padding': '10px'}),
        dcc.Graph(id='bar-chart-dash', style={'width': '50%', 'display': 'inline-block', 'padding': '10px'})
    ], style={'display': 'flex', 'flexWrap': 'wrap'}) # Use flex to arrange graphs side-by-side
])

# Define callbacks to update plots based on dropdown and radio button selections
@app_dash_example.callback(
    Output('scatter-plot-dash', 'figure'),
    Output('bar-chart-dash', 'figure'),
    Input('day-dropdown', 'value'),
    Input('smoker-radio', 'value')
)
def update_dashboard_plots(selected_day, selected_smoker):
    filtered_df = tips_dash_app_df[tips_dash_app_df.day == selected_day]

    if selected_smoker != 'All':
        filtered_df = filtered_df[filtered_df.smoker == selected_smoker]

    # Scatter plot: Total Bill vs Tip
    scatter_fig = px.scatter(filtered_df,
                             x="total_bill", y="tip",
                             color="time", size="size",
                             hover_name="sex",
                             title=f"Total Bill vs Tip on {selected_day} (Smoker: {selected_smoker})",
                             labels={"total_bill": "Total Bill ($)", "tip": "Tip ($)"})
    scatter_fig.update_layout(transition_duration=500)

    # Bar chart: Average Tip by Time (Lunch/Dinner)
    avg_tip_by_time = filtered_df.groupby('time')['tip'].mean().reset_index()
    bar_fig = px.bar(avg_tip_by_time,
                     x="time", y="tip",
                     color="time",
                     title=f"Average Tip by Time on {selected_day} (Smoker: {selected_smoker})",
                     labels={"time": "Time of Day", "tip": "Average Tip ($)"})
    bar_fig.update_layout(transition_duration=500)

    return scatter_fig, bar_fig

# Run the app
if __name__ == '__main__':
    app_dash_example.run(mode='inline', debug=False) # Use app.run() for newer Dash versions:)
```

```python
import random
import json
import numpy as np
import pyvista as pv
from pyvista import examples
from vtk.util.numpy_support import vtk_to_numpy

import dash
from dash import html # Correct import for html components
from dash import dcc # Correct import for core components
import dash_bootstrap_components as dbc # Dash Bootstrap Components for layout
from dash.dependencies import Input, Output, State

import dash_vtk
from dash_vtk.utils import presets # Presets for colormaps


random.seed(42) # Set random seed for reproducibility


def toDropOption(name):
    """Helper function to format options for Dash Dropdown."""
    return {"label": name, "value": name}


# Get point cloud data from PyVista
# examples.download_crater_topo() downloads a topographical dataset of a crater.
uniformGrid = examples.download_crater_topo()
# Extract a subset of the grid for more manageable visualization.
subset = uniformGrid.extract_subset((500, 900, 400, 800, 0, 0), (5, 5, 1))


def updateWarp(factor=1):
    """
    Warps the terrain by a scalar factor (elevation) and extracts PolyData.
    This simulates terrain deformation based on a scale factor.
    """
    # Warp the terrain geometry based on its scalar (elevation) data.
    terrain = subset.warp_by_scalar(factor=factor)
    # Extract the polygonal data (surface mesh) from the warped terrain.
    polydata = terrain.extract_geometry()
    # Flatten the points array (x, y, z coordinates) for dash-vtk.
    points = polydata.points.ravel()
    # Get the polygonal connectivity (faces) for dash-vtk.
    polys = vtk_to_numpy(polydata.GetPolys().GetData())
    # Get the elevation data (scalars) associated with the points.
    elevation = polydata["scalar1of1"]
    # Calculate the min and max elevation for color mapping.
    min_elevation = np.amin(elevation)
    max_elevation = np.amax(elevation)
    return [points, polys, elevation, [min_elevation, max_elevation]]


# Initial warp of the terrain with a factor of 1.
points, polys, elevation, color_range = updateWarp(1)

# --- Dash Application Setup ---

# Initialize the Dash application with Bootstrap external stylesheets for styling.
app = dash.Dash(__name__, external_stylesheets=[dbc.themes.BOOTSTRAP])
server = app.server # Standard way to expose the Flask server for deployment.

# Define the initial Dash VTK view component.
# This sets up a 3D rendering view for a GeometryRepresentation (our terrain mesh).
vtk_view = dash_vtk.View(
    id="vtk-view", # ID for the view, used in callbacks
    pickingModes=["hover"], # Enable hover picking for tooltip
    children=[
        dash_vtk.GeometryRepresentation(
            id="vtk-representation", # ID for the geometry representation
            children=[
                dash_vtk.PolyData(
                    id="vtk-polydata", # ID for the PolyData component
                    points=points, # Initial points data
                    polys=polys, # Initial polygonal connectivity
                    children=[
                        dash_vtk.PointData( # Data associated with points
                            [
                                dash_vtk.DataArray(
                                    id="vtk-array", # ID for the DataArray
                                    registration="setScalars", # Register as scalars for coloring
                                    name="elevation", # Name of the scalar array
                                    values=elevation, # Initial elevation values
                                )
                            ]
                        )
                    ],
                )
            ],
            colorMapPreset="erdc_blue2green_muted", # Initial colormap preset
            colorDataRange=color_range, # Initial color data range
            property={"edgeVisibility": True}, # Show edges of the mesh
            showCubeAxes=True, # Show cube axes around the model
            cubeAxesStyle={"axisLabels": ["", "", "Altitude"]}, # Label the Z-axis as Altitude
        ),
        # A separate GeometryRepresentation for displaying the picked point (a sphere).
        dash_vtk.GeometryRepresentation(
            id="pick-rep",
            actor={"visibility": False}, # Initially hidden
            children=[
                dash_vtk.Algorithm(
                    id="pick-sphere", # Sphere source for the picked point
                    vtkClass="vtkSphereSource",
                    state={"radius": 100}, # Initial radius of the sphere
                )
            ],
        ),
    ],
)

# Define the layout of the Dash application using Dash Bootstrap Components.
app.layout = dbc.Container(
    fluid=True, # Container takes full width
    style={"height": "100vh", "display": "flex", "flexDirection": "column"}, # Full viewport height, column layout
    children=[
        dbc.Row( # Row for controls (slider, dropdown, checklist)
            [
                dbc.Col( # Column for the scale factor slider
                    children=dcc.Slider(
                        id="scale-factor",
                        min=0.1, max=5, step=0.1, value=1,
                        marks={0.1: "0.1", 5: "5"},
                        className="p-2" # Add some padding
                    ),
                    width=4 # Occupy 4 columns of the 12-column grid
                ),
                dbc.Col( # Column for the colormap preset dropdown
                    children=dcc.Dropdown(
                        id="dropdown-preset",
                        options=list(map(toDropOption, presets)), # Options from dash_vtk presets
                        value="erdc_rainbow_bright", # Default colormap
                        className="p-2"
                    ),
                    width=4
                ),
                dbc.Col( # Column for the cube axes toggle checklist
                    children=dcc.Checklist(
                        id="toggle-cube-axes",
                        options=[{"label": " Show axis grid", "value": "grid"}],
                        value=[], # Initially unchecked
                        labelStyle={"display": "inline-block", "marginLeft": "10px"},
                        className="p-2"
                    ),
                    width=4
                ),
            ],
            style={"height": "12%", "alignItems": "center"}, # Align items vertically in the middle
            className="mb-3" # Margin bottom
        ),
        html.Div( # Div for the main VTK viewer
            vtk_view,
            style={"flexGrow": 1, "width": "100%", "height": "88%"} # Takes remaining height
        ),
        html.Pre( # Tooltip for picked points
            id="tooltip",
            style={
                "position": "absolute",
                "bottom": "25px",
                "left": "25px",
                "zIndex": 1,
                "color": "white",
                "backgroundColor": "rgba(0,0,0,0.7)",
                "padding": "5px",
                "borderRadius": "5px",
                "pointerEvents": "none", # Ensures tooltip doesn't block interactions
                "display": "none" # Hidden by default
            },
        ),
    ],
)


# --- Dash Callbacks ---

@app.callback(
    [
        Output("vtk-representation", "showCubeAxes"),
        Output("vtk-representation", "colorMapPreset"),
        Output("vtk-representation", "colorDataRange"),
        Output("vtk-polydata", "points"),
        Output("vtk-polydata", "polys"),
        Output("vtk-array", "values"),
        Output("vtk-view", "triggerResetCamera"), # Trigger camera reset on update
    ],
    [
        Input("dropdown-preset", "value"),
        Input("scale-factor", "value"),
        Input("toggle-cube-axes", "value"),
    ],
)
def updatePresetName(name, scale_factor, cubeAxes):
    """Callback to update terrain based on slider and dropdown selections."""
    # Re-warp the terrain based on the new scale factor
    points, polys, elevation, color_range = updateWarp(scale_factor)

    # Return updated properties for the VTK components
    return [
        "grid" in cubeAxes, # showCubeAxes is true if 'grid' is in the checklist value
        name, # New colormap preset
        color_range, # New color data range
        points, # Updated points data
        polys, # Updated polygons data
        elevation, # Updated elevation values
        random.random(), # Trigger a camera reset by providing a new random value
    ]


@app.callback(
    [
        Output("tooltip", "children"),
        Output("tooltip", "style"), # Update style to control visibility
        Output("pick-sphere", "state"), # Update sphere position
        Output("pick-rep", "actor"), # Update sphere visibility
    ],
    [
        Input("vtk-view", "clickInfo"), # Listen for click events
        Input("vtk-view", "hoverInfo"), # Listen for hover events
    ],
    [
        State("tooltip", "style") # Get current style to toggle visibility
    ]
)
def onInfo(clickData, hoverData, current_style):
    """Callback to display tooltip and sphere on hover/click."""
    info = hoverData if hoverData else clickData # Prefer hover data if available

    if info:
        # Check if the picked element is our terrain representation
        if (
            "representationId" in info
            and info["representationId"] == "vtk-representation"
        ):
            # Display detailed info in the tooltip
            tooltip_text = json.dumps(info, indent=2)

            # Update sphere position to the picked world position
            sphere_state = {"center": info["worldPosition"]}

            # Make sphere visible
            actor_visibility = {"visibility": True}

            # Make tooltip visible
            new_style = {**current_style, "display": "block"}

            return tooltip_text, new_style, sphere_state, actor_visibility

        # If picked something else or no valid pick, hide tooltip and sphere
        return "", {**current_style, "display": "none"}, {}, {"visibility": False}

    # If no hover or click data, hide tooltip and sphere
    return "", {**current_style, "display": "none"}, {}, {"visibility": False}


# Run the Dash application server.
# For Jupyter/Colab, 'mode='inline'' embeds the app output directly in the notebook.
# 'debug=True' enables hot-reloading and debug tools (set to False for production).
if __name__ == "__main__":
    app.run(debug=False, mode='inline') # Corrected to app.run() for newer Dash versions
```

```python
import os
import random

import dash
from dash import html # Correct import for html components
from dash import dcc # Correct import for core components
import dash_bootstrap_components as dbc # Dash Bootstrap Components for layout
import itk # ITK for reading medical image files (like VTI)

import dash_vtk
from dash_vtk.utils import to_volume_state # Utility to convert ITK/VTK images to dash-vtk state

import vtk # VTK library, often used by ITK internally for image processing

random.seed(42) # Set random seed for reproducibility


# --- Data Loading and Volume State Creation ---
# The original code used __file__ which is not defined in Jupyter/Colab.
# We assume 'head.vti' has been downloaded to the current working directory.
head_vti_path = "/content/dash-vtk/demos/data/head.vti"

# Load the VTI dataset using VTK's reader
# vtk.vtkXMLImageDataReader is used to read the .vti file.
reader = vtk.vtkXMLImageDataReader()
reader.SetFileName(head_vti_path)
reader.Update() # Execute the reader to load the data

# Convert the VTK dataset to a state object compatible with dash-vtk.
volume_state = to_volume_state(reader.GetOutput())

# Define sliders for interactive control of slicing and window/level.
# These sliders control the i, j, k slices and the color window/level for visualization.
sliders = {
    "Slice i": dcc.Slider(id="slider-i", min=0, max=256, value=128, step=1),
    "Slice j": dcc.Slider(id="slider-j", min=0, max=256, value=128, step=1),
    "Slice k": dcc.Slider(id="slider-k", min=0, max=95, value=47, step=1), # Max value adjusted based on typical head.vti dimensions
    "Color Level": dcc.Slider(id="slider-lvl", min=0, max=4095, value=1000, step=1),
    "Color Window": dcc.Slider(id="slider-window", min=0, max=4095, value=4095, step=1),
}

# Create a Dash Bootstrap Card to group the slider controls.
controls = dbc.Card(
    body=True,
    children=dbc.Row(
        [
            # Create a column for each slider with its label.
            dbc.Col(
                [dbc.Label(label), component],
                width="auto", # Adjust width automatically
                className="d-flex flex-column align-items-center" # Center content in column
            )
            for label, component in sliders.items()
        ],
        justify="around", # Distribute columns evenly
        className="g-0" # Remove gutter between columns
    ),
    className="mb-3" # Margin bottom for the card
)

# Define the Dash VTK View for displaying the slices.
# This view is configured for 2D slicing (camera position, view up, parallel projection).
slice_view = dash_vtk.View(
    id="slice-view",
    # Camera settings for an axial view (looking down Z-axis)
    cameraPosition=[0, 0, 1], # Looking from above
    cameraViewUp=[0, -1, 0], # Y-axis points down on screen
    cameraParallelProjection=True, # Use parallel projection for 2D slices
    background=[0.9, 0.9, 1], # Light blue background
    children=[
        # ShareDataSet allows multiple representations to use the same underlying volume data.
        dash_vtk.ShareDataSet(dash_vtk.Volume(state=volume_state)),
        # SliceRepresentation for the I-slice (sagittal)
        dash_vtk.SliceRepresentation(
            id="slice-repr-i",
            iSlice=128, # Initial slice index
            property={"colorWindow": 4095, "colorLevel": 1000}, # Initial window/level
            children=dash_vtk.ShareDataSet(), # Shares data from parent ShareDataSet
        ),
        # SliceRepresentation for the J-slice (coronal)
        dash_vtk.SliceRepresentation(
            id="slice-repr-j",
            jSlice=128, # Initial slice index
            property={"colorWindow": 4095, "colorLevel": 1000},
            children=dash_vtk.ShareDataSet(),
        ),
        # SliceRepresentation for the K-slice (axial)
        dash_vtk.SliceRepresentation(
            id="slice-repr-k",
            kSlice=47, # Initial slice index
            property={"colorWindow": 4095, "colorLevel": 1000},
            children=dash_vtk.ShareDataSet(),
        ),
    ],
)

# Initialize the Dash application with Bootstrap external stylesheets.
app = dash.Dash(__name__, external_stylesheets=[dbc.themes.BOOTSTRAP])
server = app.server # Standard way to expose the Flask server for deployment.

# Define the main layout of the Dash application.
app.layout = dbc.Container(
    fluid=True, # Container takes full width
    style={"height": "100vh", "display": "flex", "flexDirection": "column"}, # Full viewport height, column layout
    children=[
        html.Div( # Div for header and controls
            style={"height": "20%", "display": "flex", "flexDirection": "column", "alignItems": "center", "justifyContent": "center"},
            children=[
                html.H2("Interactive Medical Image Slicer", className="mb-3 text-center text-gray-800"),
                controls, # The Card containing sliders
            ],
        ),
        html.Div( # Div for the main VTK slice viewer
            slice_view,
            style={"height": "80%", "width": "100%", "flexGrow": 1} # Takes remaining height
        ),
    ],
)


# --- Dash Callbacks ---

@app.callback(
    [
        Output("slice-view", "triggerRender"), # Trigger a re-render of the VTK view
        Output("slice-repr-i", "property"), # Update properties for I-slice
        Output("slice-repr-i", "iSlice"), # Update I-slice index
        Output("slice-repr-j", "property"), # Update properties for J-slice
        Output("slice-repr-j", "jSlice"), # Update J-slice index
        Output("slice-repr-k", "property"), # Update properties for K-slice
        Output("slice-repr-k", "kSlice"), # Update K-slice index
    ],
    [
        Input("slider-i", "value"), # Input from Slice i slider
        Input("slider-j", "value"), # Input from Slice j slider
        Input("slider-k", "value"), # Input from Slice k slider
        Input("slider-lvl", "value"), # Input from Color Level slider
        Input("slider-window", "value"), # Input from Color Window slider
    ],
)
def update_slice_property(i, j, k, level, window):
    """
    Callback to update slice positions and color window/level based on slider inputs.
    A random value is returned for triggerRender to force the VTK view to update.
    """
    render_call = random.random() # Generate a random number to trigger re-render
    slice_prop = {"colorLevel": level, "colorWindow": window} # Create property dict for slices
    return render_call, slice_prop, i, slice_prop, j, slice_prop, k


# Run the Dash application server.
# For Jupyter/Colab, 'mode='inline'' embeds the app output directly in the notebook.
# 'debug=True' enables hot-reloading and debug tools (set to False for production).
if __name__ == "__main__":
    app.run(debug=False, mode='inline') # Corrected to app.run() for newer Dash versions
```

```python
import random

import dash
from dash import html # Correct import for html components
from dash import dcc # Correct import for core components
import dash_bootstrap_components as dbc # Dash Bootstrap Components for layout
from dash.dependencies import Input, Output, State # Standard Dash imports for callbacks

import dash_vtk # Dash VTK components

# vtk is not explicitly imported in the provided snippet as vtkConeSource is a vtk.js class
# represented by dash_vtk.Algorithm, but it's good to be aware it's VTK-related.
# import vtk

random.seed(42) # Set random seed for reproducibility


# Initialize the Dash application with Bootstrap external stylesheets for styling.
app = dash.Dash(__name__, external_stylesheets=[dbc.themes.BOOTSTRAP])
server = app.server # Standard way to expose the Flask server for deployment.

# Define the Dash VTK View component for displaying the 3D cone.
vtk_view = dash_vtk.View(
    id="geometry-view", # ID for the view, used in callbacks for camera reset
    children=[
        dash_vtk.GeometryRepresentation( # Represents a geometric actor in the 3D scene
            [
                dash_vtk.Algorithm( # Use dash_vtk.Algorithm to create VTK objects like vtkConeSource
                    id="vtk-algorithm", # ID for the algorithm, used in callbacks to update its state
                    vtkClass="vtkConeSource", # Specifies the VTK class to use (vtk.js equivalent)
                    state={"capping": False, "resolution": 60}, # Initial properties of the cone
                )
            ]
        )
    ],
)

# Define the control card using Dash Bootstrap Components.
# This card contains a slider for cone resolution and a switch for capping.
controls = dbc.Card(
    [
        dbc.CardHeader("Controls"), # Header for the control card
        dbc.CardBody(
            [
                html.P("Resolution:"), # Label for the resolution slider
                dcc.Slider( # Slider for controlling the cone's resolution
                    id="slider-resolution",
                    min=10, max=100, step=1, value=60, # Slider range and initial value
                    marks={10: "10", 100: "100"}, # Marks on the slider
                    className="mb-3" # Margin bottom for spacing
                ),
                html.Br(), # Line break for spacing
                dbc.Checklist( # Checklist (rendered as a switch) for capping
                    options=[{"label": "Capping", "value": "capping"}],
                    value=[], # Initial value (unchecked)
                    id="capping-checklist", # ID for the checklist
                    switch=True, # Render as a switch
                ),
            ]
        ),
    ]
)

# Define the main layout of the Dash application using Dash Bootstrap Components.
app.layout = dbc.Container(
    fluid=True, # Container takes full width
    style={"marginTop": "15px", "height": "calc(100vh - 30px)", "display": "flex", "flexDirection": "column"}, # Layout styling
    children=[
        dbc.Row( # Row for arranging controls and the VTK view
            [
                dbc.Col(width=4, children=controls, className="mb-3 mb-md-0"), # Column for controls, responsive margin
                dbc.Col( # Column for the VTK view
                    width=8,
                    children=[
                        html.Div(
                            vtk_view, # Embed the VTK view component
                            style={"height": "100%", "width": "100%", "minHeight": "400px"}, # Styling for the VTK container
                        )
                    ],
                ),
            ],
            style={"flexGrow": 1}, # Allow row to grow and take available height
            className="g-0" # Remove gutter between columns
        ),
    ],
)


# --- Dash Callbacks ---

@app.callback(
    [Output("vtk-algorithm", "state"), Output("geometry-view", "triggerResetCamera")], # Outputs: algorithm state and camera reset trigger
    [Input("slider-resolution", "value"), Input("capping-checklist", "value")], # Inputs: slider value and checklist value
)
def update_cone(slider_val, checked_values):
    """
    Callback function to update the cone's properties (resolution and capping)
    based on slider and checklist inputs.
    """
    # Create a new state dictionary for the vtkConeSource algorithm.
    # 'resolution' is set directly from the slider value.
    # 'capping' is true if 'capping' is present in the checked_values list.
    new_state = {"resolution": slider_val, "capping": "capping" in checked_values}

    # Return the new state for the algorithm and a random number to trigger a camera reset.
    # Triggering camera reset ensures the view updates correctly after geometry changes.
    return new_state, random.random()


# Run the Dash application server.
# For Jupyter/Colab, 'mode='inline'' embeds the app output directly in the notebook.
# 'debug=True' enables hot-reloading and debug tools (set to False for production).
if __name__ == "__main__":
    app.run(debug=False, mode='inline') # Corrected to app.run() for newer Dash versions
```

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns # For loading a dataset and potentially for cleaner defaults
%matplotlib inline

# Simulate data for User Acquisition Channels
np.random.seed(20)
months_ua = pd.date_range(start='2023-01-01', periods=18, freq='M')
channels = ['Organic Search', 'Paid Ads', 'Social Media', 'Email Marketing', 'Referral', 'Direct']

# Generate user data with some general trends and one "problematic" channel
ua_data = {}
for i, channel in enumerate(channels):
    base_users = 500 + i * 50 + np.random.normal(0, 50, len(months_ua))
    # Introduce a dip in 'Social Media' performance
    if channel == 'Social Media':
        base_users[8:12] -= 200 # Significant drop
        base_users[12:] += 150 # Recovery

    # Introduce a sharp spike in 'Paid Ads'
    if channel == 'Paid Ads':
        base_users[10] += 500 # A sudden, unusual spike
        base_users[11] += 300

    ua_data[channel] = np.maximum(0, base_users).astype(int) # Ensure no negative users

df_ua = pd.DataFrame(ua_data, index=months_ua)
df_ua.index.name = 'Month'
df_ua_long = df_ua.reset_index().melt(id_vars='Month', var_name='Channel', value_name='Users')

print("Sample User Acquisition Data:")
print(df_ua_long.head())

# --- "Bad" Chart (Matplotlib default, overlapping lines) ---
plt.figure(figsize=(14, 8))

for channel in channels:
    plt.plot(df_ua.index, df_ua[channel], label=channel, linewidth=1.5, alpha=0.8)

plt.title('Monthly User Acquisition Across All Channels', fontsize=18, fontweight='bold')
plt.xlabel('Month', fontsize=14)
plt.ylabel('Number of Users', fontsize=14)
plt.xticks(rotation=45, ha='right', fontsize=10)
plt.yticks(fontsize=10)
plt.legend(loc='upper left', bbox_to_anchor=(1, 1), ncol=1, fontsize=10)
plt.grid(True, linestyle='-', alpha=0.3)
plt.tight_layout()
plt.show()


# --- "Good" Chart (Seaborn & Matplotlib for Focused Storytelling) ---
plt.figure(figsize=(12, 7))

# Set a professional Seaborn style
sns.set_style("whitegrid")
sns.set_palette("muted") # A more subdued palette for background channels

# Highlight specific channels and de-emphasize others
highlight_channel_1 = 'Social Media' # Channel with a dip and recovery
highlight_channel_2 = 'Paid Ads'     # Channel with a spike

for channel in channels:
    if channel == highlight_channel_1:
        sns.lineplot(x=df_ua.index, y=df_ua[channel], label=highlight_channel_1,
                     color='darkorange', linewidth=3, marker='o', markersize=7, zorder=3, ax=plt.gca())
    elif channel == highlight_channel_2:
        sns.lineplot(x=df_ua.index, y=df_ua[channel], label=highlight_channel_2,
                     color='darkviolet', linewidth=3, marker='X', markersize=7, linestyle='-', zorder=3, ax=plt.gca())
    else:
        sns.lineplot(x=df_ua.index, y=df_ua[channel], color='gray', linestyle=':', linewidth=1, alpha=0.3, zorder=1, ax=plt.gca())

# Add annotations for key insights
plt.annotate('Social Media dip & recovery',
             xy=(df_ua.index[10], df_ua.loc[df_ua.index[10], highlight_channel_1]),
             xytext=(df_ua.index[12], df_ua.loc[df_ua.index[10], highlight_channel_1] - 150),
             arrowprops=dict(facecolor='darkorange', shrink=0.05, width=1, headwidth=6),
             fontsize=10, color='darkorange', fontweight='bold',
             bbox=dict(boxstyle="round,pad=0.3", fc="white", ec="darkorange", lw=0.8, alpha=0.9))

plt.annotate('Unusual spike in Paid Ads!',
             xy=(df_ua.index[10], df_ua.loc[df_ua.index[10], highlight_channel_2]),
             xytext=(df_ua.index[7], df_ua.loc[df_ua.index[10], highlight_channel_2] + 200),
             arrowprops=dict(facecolor='darkviolet', shrink=0.05, width=1, headwidth=6),
             fontsize=10, color='darkviolet', fontweight='bold',
             bbox=dict(boxstyle="round,pad=0.3", fc="white", ec="darkviolet", lw=0.8, alpha=0.9))

plt.title('User Acquisition: Highlighting Social Media & Paid Ads Performance', fontsize=18, fontweight='bold', color='#2c3e50')
plt.xlabel('Month', fontsize=14, color='#555')
plt.ylabel('Number of Users', fontsize=14, color='#555')
plt.xticks(rotation=45, ha='right', fontsize=10)
plt.yticks(fontsize=10)
plt.legend(loc='upper left', bbox_to_anchor=(1.02, 1), ncol=1, fontsize=10, frameon=False)
plt.grid(axis='y', linestyle=':', alpha=0.5)

plt.tight_layout(rect=[0, 0, 0.88, 1])
plt.show()
```

# **Data Storytelling: The Art of Impact ✔️**

## Creating beautiful plots is just the **first** step. The **true power of data visualization** lies in **transforming complex analyses into compelling stories that drive understanding and action.** This section highlights key principles to make your visualizations **impactful.**


```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns # For loading a dataset and potentially for cleaner defaults
%matplotlib inline
```

```python
# Simulate data for demonstration: Monthly sales for several products
np.random.seed(42)
months = pd.date_range(start='2023-01-01', periods=12, freq='ME')
products = [f'Product {i+1}' for i in range(7)] # 7 products

# Generate sales data with some trends and noise
sales_data = {}
for i, product in enumerate(products):
    base_sales = 100 + i * 10 # Different base for each product
    trend = np.linspace(0, 50, 12) # General upward trend
    noise = np.random.normal(0, 15, 12) # Random fluctuations
    seasonal = 10 * np.sin(np.linspace(0, 2 * np.pi, 12)) # Seasonal pattern
    sales_data[product] = base_sales + trend + noise + seasonal
    # Make one product (e.g., Product 4) consistently higher/lower
    if product == 'Product 4':
        sales_data[product] += 50 # Make it a standout
    elif product == 'Product 2':
        sales_data[product] -= 30 # Make it a low performer

df_sales = pd.DataFrame(sales_data, index=months)
df_sales.index.name = 'Month'
df_sales_long = df_sales.reset_index().melt(id_vars='Month', var_name='Product', value_name='Sales')

print("Sample Sales Data:")
print(df_sales_long.head())
```

# **1. The "Bad" Chart: Clutter and Confusion 👎**
## This chart is designed to demonstrate **common pitfalls** in data visualization. It's **cluttered**, uses default settings that don't highlight insights, and makes it **hard** to understand the main message.


```python
# --- "Bad" Chart ---
plt.figure(figsize=(12, 7))

# Plot all products with default Matplotlib colors
for product in products:
    plt.plot(df_sales.index, df_sales[product], label=product, linewidth=1, alpha=0.8)

plt.title('Monthly Sales Performance of All Products (Cluttered)', fontsize=16, fontweight='bold')
plt.xlabel('Month', fontsize=12)
plt.ylabel('Sales Volume', fontsize=12)
plt.xticks(rotation=45, ha='right')
plt.legend(loc='upper left', bbox_to_anchor=(1, 1), ncol=1, fontsize=9) # Legend outside, but still busy
plt.grid(True, linestyle='-', alpha=0.3) # Heavy grid lines
plt.tight_layout()
plt.show()
```

# **Issues with the "Bad" Chart:**

## **Too many lines:** Hard to track individual product performance.
## **Generic colors:** No clear distinction or emphasis.
## **Busy legend:** Takes up space and doesn't immediately convey importance.
## **Heavy grid:** Adds visual noise.
## **Lack of clear message:** What's the main takeaway?

# **2. The "Good" Chart: Clarity and Impact 👍**

## Now, let's apply data storytelling principles to the same data. We'll focus on a **clear message**, **strategic use of color**, and **minimizing chartjunk** to make the insights immediately apparent.


```python
# --- "Good" Chart ---
plt.figure(figsize=(12, 7)) # Slightly larger figure for more detail

# Set a clean Seaborn style for better aesthetics
sns.set_style("whitegrid")
sns.set_palette("viridis") # A visually appealing palette

# Highlight key products and subtly show others
highlight_product_1 = 'Product 4'
highlight_product_2 = 'Product 2' # Second product to highlight (e.g., worst performer)

for product in products:
    if product == highlight_product_1:
        # Highlight best performing product
        sns.lineplot(x=df_sales.index, y=df_sales[product], label=highlight_product_1,
                     color='darkgreen', linewidth=3.5, marker='o', markersize=8, zorder=3, ax=plt.gca())
    elif product == highlight_product_2:
        # Highlight worst performing product
        sns.lineplot(x=df_sales.index, y=df_sales[product], label=highlight_product_2,
                     color='darkred', linewidth=2.5, marker='X', markersize=7, linestyle='--', zorder=2, ax=plt.gca())
    else:
        # Subtly show other products as context
        sns.lineplot(x=df_sales.index, y=df_sales[product], color='gray', linestyle=':', linewidth=1, alpha=0.4, zorder=1, ax=plt.gca())

# Add a shaded region to highlight a specific period (e.g., a marketing campaign)
# Let's say a campaign ran from April to July (index 3 to 6)
plt.axvspan(df_sales.index[3], df_sales.index[6], color='blue', alpha=0.08, label='Q2 Campaign')


plt.title(f'Product Performance: Highlighting Leaders & Laggards (2023)', fontsize=18, fontweight='bold', color='#2c3e50')
plt.xlabel('Month', fontsize=14, color='#555')
plt.ylabel('Sales Volume', fontsize=14, color='#555')
plt.xticks(rotation=45, ha='right', fontsize=10)
plt.yticks(fontsize=10)
plt.ylim(0, df_sales.max().max() + 80) # Adjust y-limit for better spacing

# Custom legend to order and emphasize highlighted products
handles, labels = plt.gca().get_legend_handles_labels()
# Reorder legend to put highlighted products first
order = [labels.index(highlight_product_1), labels.index(highlight_product_2)] + \
        [i for i, label in enumerate(labels) if label not in [highlight_product_1, highlight_product_2, 'Q2 Campaign']] + \
        [labels.index('Q2 Campaign')]
plt.legend([handles[idx] for idx in order], [labels[idx] for idx in order],
           loc='upper left', bbox_to_anchor=(1.02, 1), ncol=1, fontsize=10, frameon=False) # Clean legend, no frame

# Add annotations for key insights
plt.annotate(f'{highlight_product_1} consistently leads!',
             xy=(df_sales.index[10], df_sales[highlight_product_1].iloc[10]),
             xytext=(df_sales.index[7], df_sales[highlight_product_1].iloc[10] + 40),
             arrowprops=dict(facecolor='darkgreen', shrink=0.05, width=1, headwidth=6),
             fontsize=11, color='darkgreen', fontweight='bold',
             bbox=dict(boxstyle="round,pad=0.3", fc="white", ec="darkgreen", lw=0.8, alpha=0.9))

plt.annotate(f'{highlight_product_2} lags behind.',
             xy=(df_sales.index[8], df_sales[highlight_product_2].iloc[8]),
             xytext=(df_sales.index[5], df_sales[highlight_product_2].iloc[8] - 50),
             arrowprops=dict(facecolor='darkred', shrink=0.05, width=1, headwidth=6),
             fontsize=11, color='darkred', fontweight='bold',
             bbox=dict(boxstyle="round,pad=0.3", fc="white", ec="darkred", lw=0.8, alpha=0.9))

plt.tight_layout(rect=[0, 0, 0.88, 1]) # Adjust layout to make space for legend and annotations
plt.show()
```

# **Key Takeaways from the "Good" Chart:**

## **Clear Message:** The title immediately tells the story.

## **Strategic Color:** "Product 4" stands out, while other products provide context without clutter.

## **Minimal Chartjunk:** Clean grid, no unnecessary elements.

## **Narrative Flow:** The annotation guides the viewer to a key insight.

# This transformation demonstrates how **intentional design choices** can elevate your data visualizations from **mere plots to powerful storytelling tools, driving actionable insights and influencing decisions.**

# **3. Book Recommendations: Deepen Your Understanding 📖**
## Dive deeper into the art and science of data visualization with these highly recommended books.


#  **📕"Storytelling with Data" by Cole Nussbaumer Knaflic**
## **Focus:** Essential for effective communication through data. Teaches you how to craft compelling narratives with your visualizations, focusing on clarity, simplicity, and impact.
## **Key Takeaway:** It's not just about making a chart; it's about telling a story that drives action.

# **📕"Python for Data Analysis" by Wes McKinney**
## **Focus:** A foundational text for data manipulation and analysis in Python. Written by the creator of the `pandas` library.
## **Key Takeaway:** Master data cleaning, transformation, and preparation, which are crucial steps before visualization.

# **📗"Grokking Algorithms" by Aditya Bhargava (Manning Publications)**
## **Focus:** An incredibly intuitive and visual guide to common algorithms and data structures. While not strictly about data visualization, understanding algorithms is fundamental to what you visualize.
## **Key Takeaway:** Makes complex algorithms easy to grasp with engaging illustrations and practical examples. Excellent for building foundational computer science knowledge.

# **"📗Python for Data Science Handbook" by Jake VanderPlas (O'Reilly)**
## **Focus:** A comprehensive reference for essential tools in data science, including NumPy, Pandas, Matplotlib, and Scikit-Learn.
## **Key Takeaway:** Provides a deep dive into the core libraries for data manipulation, scientific computing, and visualization, offering practical examples and best practices. A must-have for any Python data scientist.

# **"📗Fundamentals of Data Visualization" by Claus O. Wilke (O'Reilly)**
## **Focus:** Explores the principles of good data visualization from a theoretical and practical perspective, covering design choices, common pitfalls, and effective presentation.
## **Key Takeaway:** Teaches you how to create clear, accurate, and compelling visualizations by understanding the underlying principles of visual perception and effective design.

# **"📘Data Visualization with Category Theory and Geometry" by Lukas Silvester Barth et al.**
## **Focus:** A highly academic and advanced book that explores the deep mathematical principles behind data visualization. It delves into a critical analysis and refinement of algorithms like UMAP.
## **Key Takeaway:** Provides a systematic mathematical foundation for data representation. It's a great resource for researchers and those who want to understand the theoretical underpinnings of visualization and improve upon existing algorithms.

# **4. Personal Projects: Learn by Doing 🚀**
## The best way to solidify your learning is by applying your skills to real-world problems you care about.


### **Start with data you're passionate about!** Whether it's sports statistics, movie ratings, climate data, or social trends, find data that genuinely interests you.
### **Build small projects:** Don't aim for perfection initially. Focus on completing small, functional projects.
### **Share your work:** Put your projects on GitHub, write about them, and seek feedback. This is invaluable for growth.
### **Contribute to open source:** Look for opportunities to contribute to existing data visualization libraries or projects.


## **Your journey to becoming a data visualization expert is an exciting one. Keep exploring, keep building, and keep telling impactful stories with data! Good luck 🤗✨🚀📈**
